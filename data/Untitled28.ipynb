```python
import pandas as pd
import json
```

```python
lexer_table = pd.read_csv('lexer_table.csv', index_col=0).fillna(0)
```

```python
bool_columns = ['ACCEPT', 'STACK', 'ERROR', 'RETURN']
int_columns = ['NEXT']
follow_symbols_column = 'FOLLOW_SYMBOLS'
```

```python
TOKENS_DICT = {
    # Arg
    'FUNCTION_NAME': 'FUNCTION_NAME',
    'BOOL': 'BOOL',
    'NUMERIC': 'NUMERIC',
    'SEPARATOR': ',',
    # Brackets
    'LPAR': '(',
    'RPAR': ')',
    # Ariphmetics
    'MUT': '*',
    'REMUT': '/',
    'SUM': '+',
    'DELTA': '-',
    # Boolean
    'AND': '&',
    'OR': '|',
    'NOT': '!',
    # Compear
    'LARGER': '>',
    'BIGGER': '<',
    'BEQUAL': '=',
    # Specific
    'if': 'if',
    'resolve': 'resolve',
    'foreach': 'foreach',
    'equal': 'equal',
    # Eps
    'eps': 'eps'
}
RETOKENS_DICT = {
    j: i for i, j in TOKENS_DICT.items()
}
```

```python
processsed_lexer_table = lexer_table.loc[:, bool_columns].astype('bool').fillna(False)
processsed_lexer_table.loc[:, int_columns] = lexer_table.loc[:, int_columns].astype('uint')
processsed_lexer_table.loc[:, follow_symbols_column] = lexer_table.loc[:, ['FOLLOW_SYMBOLS']].map(lambda a: [RETOKENS_DICT[i] for i in a.split(' ')])
processsed_lexer_table['NEXT'] = processsed_lexer_table['NEXT'].map(lambda a: a if a > 0 else pd.NA)
processsed_lexer_table.to_csv('processsed_lexer_table.csv')
```
